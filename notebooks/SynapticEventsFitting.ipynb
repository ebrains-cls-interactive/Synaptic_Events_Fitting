# Synaptic Events Fitting

---

# A. Prepare configuration

```python
from synapticwidgets.api import SynapticEventsFitting
from synapticwidgets.api import prepare_workspace_environment

paths = prepare_workspace_environment() # Prepare workspace environment for NSG 

workflow = SynapticEventsFitting(data_path=paths["data_path"],transfer_path=paths["transfer_path"])
display(workflow)
```

---

# B. Run simulation via NSG

```python
from synapticwidgets.api import RunOnNSG

workflow = RunOnNSG(data_path=paths["data_path"],transfer_path=paths["transfer_path"])
display(workflow)
```

---

# C. Run simulation via Service Account

```python
from synapticwidgets.api import RunWithSA

workflow = RunWithSA(data_path=paths["data_path"],transfer_path=paths["transfer_path"])
display(workflow)
```
