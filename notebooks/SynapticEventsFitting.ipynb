# Synaptic Events Fitting

---

# A. Prepare configuration

```python
from clsiwidgets.api import SynapticEventsFitting
from clsiwidgets.api import prepare_workspace_environment

paths = prepare_workspace_environment() # Prepare workspace environment for NSG 

workflow = SynapticEventsFitting(data_path=paths["data_path"],transfer_path=paths["transfer_path"])
display(workflow)
```

---

# B. Run simulation via NSG

```python
from clsiwidgets.api import RunOnNSG

workflow = RunOnNSG(data_path=paths["data_path"],transfer_path=paths["transfer_path"], results_path=paths["results_path"])
display(workflow)
```

---

# C. Run simulation via Service Account

```python
from clsiwidgets.api import RunWithSA

workflow = RunWithSA(data_path=paths["data_path"],transfer_path=paths["transfer_path"], results_path=paths["results_path"])
display(workflow)
```

---

# D. Analyze simulation

```python
from clsiwidgets.api import AnalyzeSimulationResults

workflow = AnalyzeSimulationResults(results_path=paths["results_path"], best_fit_workspace_path=paths["best_fit_workspace_path"])
display(workflow)
```
